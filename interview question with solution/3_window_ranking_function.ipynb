### Window function
- ##### row_number()
- ##### rank()
- ##### dense_rank()

|Name|Subject|Marks|row_number|rank|dense_rank|
|---|---|---|---|---|---|
|Ravi|Computer|99|1|1|1|
|Jitu|Computer|99|2|1|1|
|Ramesh|Computer|98|3|3|2|
|Ankit|Computer|91|4|4|3|
|Shani|Computer|88|5|5|4|
|Ajay|Computer|86|6|6|5|

| Function       | Ties              | Gaps? |
|----------------|-------------------|-------|
| ROW_NUMBER()   | Different numbers | N/A   |
| RANK()         | Same rank         | Yes   |
| DENSE_RANK()   | Same rank         | No    |

```python
data = [
    ('Ajay', 'Computer', 86),
    ('Ravi', 'Computer', 99),
    ('Shani', 'Computer', 88),
    ('Ramesh', 'Computer', 98),
    ('Ankit', 'Computer', 91),
    ('Jitu', 'Computer', 99),
]
columns =['Name', 'Subject', 'Marks']

df = spark.createDataFrame(data, columns)
df.display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, rank, dense_rank, col

win_spec = Window.partitionBy('Subject').orderBy(col('Marks').desc())
rank_df = df.withColumn('row_number', row_number().over(win_spec))\
            .withColumn('rank', rank().over(win_spec))\
            .withColumn('dense_rank', dense_rank().over(win_spec))
rank_df.display()
```
