### 1. Remove duplicates from a dataset containing customer data.

|customer_id|name|email|city|salary|
|---|---|---|---|---|
|1|Alice|alice@example.com|New York|85000|
|2|Bob|bob@example.com|Chicago|72000|
|3|Charlie|charlie@example.com|Boston|64000|
|2|Bob|bob@example.com|Chicago|72000|
|4|Diana|diana@example.com|Seattle|91000|
|3|Charlie|charlie@example.com|Boston|64000|
|5|Eve|eve@example.com|Austin|58000|
|1|Alice|alice@example.com|New York|85000|
|6|Frank|frank@example.com|Denver|76000|
|2|Bob|bob@example.com|Chicago2|72000|

```python

data = [
    (1, "Alice", "alice@example.com", "New York", 85000),
    (2, "Bob", "bob@example.com", "Chicago", 72000),
    (3, "Charlie", "charlie@example.com", "Boston", 64000),
    (2, "Bob", "bob@example.com", "Chicago", 72000),      # duplicate of customer_id 2
    (4, "Diana", "diana@example.com", "Seattle", 91000),
    (3, "Charlie", "charlie@example.com", "Boston", 64000), # duplicate of customer_id 3
    (5, "Eve", "eve@example.com", "Austin", 58000),
    (1, "Alice", "alice@example.com", "New York", 85000), # duplicate of customer_id 1
    (6, "Frank", "frank@example.com", "Denver", 76000),
    (2, "Bob", "bob@example.com", "Chicago2", 72000), # another duplicate of customer_id 2
]

columns = ["customer_id", "name", "email", "city", "salary"]
```

```python
customer_df = spark.createDataFrame(data, columns)
customer_df.show()
```

```python
org_df = customer_df.dropDuplicates()
```

```python
org_df.show()
```

```python
customer_df.dropDuplicates(['email']).display()
```

```python
customer_df.dropDuplicates(['city']).display()
```
