### Find the Second-highest salary in  each department

```python

data = [
    (1, "Alice", "alice@example.com", "Engineering", 85000),
    (2, "Bob", "bob@example.com", "Sales", 72000),
    (3, "Charlie", "charlie@example.com", "Engineering", 64000),
    (4, "Diana", "diana@example.com", "Marketing", 91000),
    (5, "Eve", "eve@example.com", "Marketing", 58000),
    (6, "Diana", "diana@example.com", "Engineering", 95000),
    (7, "Diana", "diana@example.com", "Marketing", 90000),
]

columns = ["employee_id", "name", "email", "department", "salary"]
emp_df = spark.createDataFrame(data, columns)
emp_df.display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import dense_rank, col

wind_spec = Window.partitionBy("department").orderBy(col("salary").desc())
scond_hs = emp_df.withColumn('rk', dense_rank().over(wind_spec))
scond_hs.show()
```

```python
scond_hs.filter(col("rk")==3).drop("rk").display()
```
