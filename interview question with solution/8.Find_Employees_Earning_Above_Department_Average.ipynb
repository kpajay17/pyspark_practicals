# Find Employees Earning Above Their Department Average

## Business Scenario

A large IT company wants to identify employees whose salary is higher than the average salary of their department.

## Why the Business Needs This

- Identify high-cost employees
- Review salary fairness within departments
- Detect salary anomalies
- Plan budget and compensation reviews
- Prepare promotion and retention strategies


## Sample Employee Data

| emp_id | emp_name | department | salary |
|--------|----------|------------|--------|
| 101    | Amit     | IT         | 90000  |
| 102    | Rahul    | IT         | 70000  |
| 103    | Neha     | IT         | 80000  |
| 104    | Priya    | HR         | 60000  |
| 105    | Mohit    | HR         | 50000  |
| 106    | Anjali   | HR         | 55000  |
| 107    | Vikas    | Finance    | 120000 |
| 108    | Sneha    | Finance    | 100000 |
| 109    | Karan    | Finance    | 90000  |

## Expected Output

| emp_id | emp_name | department | salary | dept_avg_salary |
|--------|----------|------------|--------|-----------------|
| 101    | Amit     | IT         | 90000  | 80000           |
| 104    | Priya    | HR         | 60000  | 55000           |
| 107    | Vikas    | Finance    | 120000 | 103333          |

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, col
from pyspark.sql.window import Window


data = [
    (101,"Amit","IT",90000),
    (102,"Rahul","IT",70000),
    (103,"Neha","IT",80000),
    (104,"Priya","HR",60000),
    (105,"Mohit","HR",50000),
    (106,"Anjali","HR",55000),
    (107,"Vikas","Finance",120000),
    (108,"Sneha","Finance",100000),
    (109,"Karan","Finance",90000)
]

df = spark.createDataFrame(
    data,
    ["emp_id","emp_name","department","salary"]
)

df.display()
```

```python
window_spec = Window.partitionBy("department")
```

```python
dept_avg_salary = (
    df.withColumn(
        "dept_avg_salary",
        avg("salary").over(window_spec)
    )
)

dept_avg_salary.show(truncate=False)
```

```python
result = dept_avg_salary.filter(col("salary") > col("dept_avg_salary"))

result.show(truncate=False)
```
