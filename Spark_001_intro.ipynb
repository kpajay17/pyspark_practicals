```python
data = [
        [1,'ajay', 5000],
        [1,'bala', 5000],
        [1,'ajay', 5000],
        [1, 'bala', 6000],
        [4,'dinesh', 8000],
        [5,'edward', 9000],
        [6,'fiona', 11000],
        [7,'gagan', 11000],
        [8,'harry', 12000],
        [9,'ian', 13000],
        [10,'jack', 14000],
        [11,'karen', None],
        [12,'lisa', 16000],
        [13,None, 17000],
        [None,'nancy', 18000],
        [15,'', 19000],
        ['','peter', 20000],
        [17,'queen', None],
        [18,'roger', 22000],
        [19,'sam', 23000],
        [20,'tom', 24000],
        [None,None, None],
        [None,'', None],
        ['',None, None],
        ]

columns = ['id','name','salary']
```

# Creating a DataFrame from python list

```python
df = spark.createDataFrame(data,columns)
df.display()
```

```python
df.printSchema()
```

### distinct
### dropDuplicates()
### na.drop() | dropna()
### na.fill() | fillna()
### replace()


```python
# total rows colunt
display(df.count())
# display new line 
print('\n')
# count distinct rows
display(df.distinct().count())
print('\n')

# showing all distinct rows
df.distinct().show(5)
```

```python
# drop duplicate rows
df.dropDuplicates().show()
```

```python
# drop duplicate rows on specific column
df.dropDuplicates(subset=['id','name']).show()
```

```python
# remove rows containing null values in any columns
df.dropna().show()
```

```python
# remove rows if all columns have null value
df.dropna(how='all').display() # default how='any'
```

```python
# removing na from specific columns
df.dropna(subset=['name','salary']).display()
```

```python
# remove rows if all specific columns have null value in row
df.dropna(subset=['name','salary'], how='all').display()
```

```python
# remove rows containing null values | specific columns have null value
df.dropna(subset=['name','salary'], how='all').display()
```

```python
# fill na with some specific value
# df.fillna(10) 
# default subset=None, how='any', axis=0 | It will replace all integer columns with 10, because 10 is an integer value
df.fillna(10).show()
```

```python
# fill na with some specific value(string) | all string columns na value with 'unknown'
df.fillna('unknown').show() 
# fill na with some specific value for specific columns
df.fillna({'name':'unknown', 'salary':0}).show() # mixed type specific to columns
```

```python
# replace blank with some value
df.replace('', 'not_defined').show()
# replace blank with some value for specific columns
df.replace('', 'not_defined', subset=['name','salary']).show()
```
