```python
# create two dataframe to perform join operations on employee and department tables
employee = spark.createDataFrame([
    ("1", "Michael", 3000, "USD", "Sales", "New York"), 
    ("2", "Andy", 4500, "USD", "Sales", "Australia"), 
    ("3", "ML", 10000, "USD", "Sales", "Chicago"), 
    ("4", "Faraz", 12500, "USD", "Sales", "Chicago"),
    ("5", "Ravi", 15000, "USD", "Marketing", "New York"),
    ("6", "Raj", None, "USD", "Marketing", "New South Wales"),
    ("7", "Sara", 20000, "USD", "HR", "New Delhi"),
    ("8", "Priya", 22500, "USD", "HR","Chennai"),
    ("9", "Priya", None, "USD", "Tech","North Carolina"),
    ("10", "Priya", 22500, "USD", "Tech", "Noida"),
    ("11", "Priya", 22500, "USD", "Tech", "Mumbai"),
    ]
    , ["id", "name", "salary", "salary_type", "department", "city"])

department = spark.createDataFrame([("Sales", "New York"), ("Marketing", "Chicago")], ["department", "city"])
```

```python
employee.show()
department.show()
```

### Supported join types include: 
 - 'inner', 
 - 'outer', 
 - 'full', 'fullouter', 'full_outer', 
 - 'left', 'leftouter', 'left_outer', 
 - 'right', 'rightouter', 'right_outer', 
 - 'semi', 'leftsemi', 'left_semi', 
 - 'anti', 'leftanti', 'left_anti', 
 - 'cross'.

```python
employee.join(department, employee.department==department.department, how='inner').show()
```

```python
# if both tables have same column name, we can use the column name directly inside ""
employee.join(department, "department").show()
```

```python
# we can use multiple columns for join
employee.join(department, ["department", "city"], how='inner').show()
```

```python
# we can use sorting to get the join order
from pyspark.sql.functions import desc
employee.join(department, "department").sort(desc("salary")).show()
```

```python
employee.join(department, employee.department==department.department, how='left').show()
```

```python
employee.join(department, employee.department==department.department, how='right').show()
```

```python
employee.join(department, employee.department==department.department, how='leftanti').show()
```

```python
employee.join(department, employee.department==department.department, how='semi').show()
```

```python
employee.join(department, "department", how="cross").show()
```
