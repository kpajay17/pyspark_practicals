```python
# create two dataframe to perform aggregation operations
emp = spark.createDataFrame([
    ("1", "Michael", 3000, "USD", "Sales", "New York", '2023-01-01'), 
    ("2", "Andy", 4500, "USD", "Sales", "Australia", '2019-01-01'), 
    ("3", "ML", 10000, "USD", "Sales", "Chicago", '2020-12-14'), 
    ("4", "Faraz", 12500, "USD", "Sales", "Chicago", '2018-03-25'),
    ("5", "Ravi", 15000, "USD", "Marketing", "New York", '2022-01-01'),
    ("6", "Raj", None, "USD", "Marketing", "New South Wales", '2023-01-01'),
    ("7", "Sara", 20000, "USD", "HR", "New Delhi", '2023-11-05'),
    ("8", "Priya", 22500, "USD", "HR","Chennai", '2023-02-19'),
    ("9", "Priya", None, "USD", "Tech","North Carolina", '2023-12-12'),
    ("10", "Priya", 22500, "USD", "Tech", "Noida", '2023-01-12'),
    ("11", "Priya", 22500, "USD", "Tech", "Mumbai", '2026-12-12'),
    ]
    , ["id", "name", "salary", "salary_type", "department", "city", 'doj'])

# department = spark.createDataFrame([("Sales", "New York"), ("Marketing", "Chicago")], ["department", "city"])
emp.show()
```

```python
emp.printSchema()
```

```python
df1 = emp.groupBy('department').sum('salary')
df1.show()
```

```python
# perform multiple aggrgate function
from pyspark.sql.functions import sum, max, min
df = emp.groupBy('department').agg(sum('salary').alias('total_sal'), max('salary').alias('max_sal'), min('salary').alias('min_sal'))
df.show()
```

```python
# grouping on multiple columns
emp.groupBy('department', 'city').sum('salary').display()
```

```python
# nesting of groupBy is not possible
df.groupBy('department').groupBy('city').sum('salary').display()
```
