```python
data = [
        [1,'ajay', 5000],
        [1,'bala', 5000],
        [1,'ajay', 5000],
        [1, 'bala', 6000],
        [4,'dinesh', 8000],
        [5,'edward', 9000],
        [6,'fiona', 11000],
        [7,'gagan', 11000],
        [8,'harry', 12000],
        [9,'ian', 13000],
        [10,'jack', 14000],
        [11,'karen', None],
        [12,'lisa', 16000],
        [13,None, 17000],
        [None,'nancy', 18000],
        [15,'', 19000],
        ['','peter', 20000],
        [17,'queen', None],
        [18,'roger', 22000],
        [19,'sam', 23000],
        [20,'tom', 24000],
        [None,None, None],
        [None,'', None],
        ['',None, None],
        ]

columns = ['id','name','salary']
df = spark.createDataFrame(data,columns)
df.show()
```

## Select
- #### syntax
  select(*cols: Union[Sequence[ForwardRef('ColumnOrName')], ForwardRef('ColumnOrName')])
  - use to select one or multiple columns

```python
# selecting column from df using column name as 'string'
df.select('name').show(3)
# selecting multiples column
df.select('name', 'salary').show(3)
```

```python
# selecting rows using columns as column object
from pyspark.sql.functions import col
df.select(col('name')).show(3)
# selecting multiples columns
df.select(col('name'), col('salary')).show(3)
# another approache
df.select(df.name, df.salary).show(3)
```

### selectExpr
- #### syntax
        selectExpr(*expr: Union[str, List[str]])
    - it will evaluate the expression inside string and return the result

```python
df.selectExpr('id', 'name', 'salary', 'salary*1.10 as inc_salary').show(3)
```

```python
from pyspark.sql.functions import expr
df.select('name', expr('salary*1.10 as inc_salary')).show(3)
```

## filter
- #### syntax
        df.filter(condition: Union[pyspark.sql.column.Column, str])
    -   use to filter out the rows based on provied condition

```python
df.filter(df.salary<=7000).show(5)
```

```python
# selecting columns and filtering
df.select('name', 'salary').filter(df.salary<=7000).show(5)
```

```python
# multiple filters
from pyspark.sql.functions import length
df.filter((df.salary<=7000) & (length(df.name)==4)).show(5)
```

## withColumn
- #### syntax
        df.withColumn(colName: str, col: pyspark.sql.column.Column)

- adding or updating column value with new value
- it will add/update one column at a time
- by nested this we can add/update multiple columns

```python
# adding new column with literal value
from pyspark.sql.functions import lit, when
df.withColumn('isactive', lit(True)).show(4)

# updating existing column 
df.withColumn('salary', df.salary*1.30).show(4)
```

## case statements
- #### syntax   
        when(condition: pyspark.sql.column.Column, value: Any).otherwise(value: Any)
- to perform case statement we use .when().otherwise() in pyspark
- we can loop .when() - .when().when().when().otherwise()
- otherwise() is mandatory
- withColumn with case statement

```python
# adding column based on condition 
df = df.withColumn('salary_type', when(df.salary>15000, 'high salary').when(df.salary>7000, 'regular_salary').otherwise('low_salary'))
df.show(5)
```

```python
df = df.withColumn('department', when(df.salary>15000, 'Tech').when(df.salary>10000, 'Finance').when(df.salary>7000, 'HR').otherwise('Marketing'))
df.show()
```

### withColumnRenamed
- #### syntax
        withColumnRenamed(existing: str, new: str)
- Used to rename the existing column
- it rename one column at a time
- we can use nested withColumnRenamed to rename multiple columns

```python
df.withColumnRenamed('salary_type', 'salary_level').show(5)
```
